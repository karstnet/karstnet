# Read and write .pline files
.pline files are text files format for Aspen SKUA (previously Gocad SKUA)

The function `networkx_to_pline` exports 


```python
import karstnet as kn
import networkx as nx
import numpy as np
```

```python
#load graph data

G = kn.io_func.networkx_from_yaml('../data/047_Seefeldhoele_000/047_Seefeldhoele_000.yaml')
```

```python

export_path = "../data/test_export_examples/"
cavename = 'Seefeldhoele'


# export the edges
kn.io_func.networkx_to_pline(G, data_type = 'lines',
                   properties = ['csdim','connected_component_number'],                       
                    name = f'{export_path}{cavename}-lines'
                    )

# export the nodes
kn.io_func.networkx_to_pline(G, data_type = 'points', 
                   properties = ['csdim'],
                    name = f'{export_path}{cavename}-points'
                    )
```
