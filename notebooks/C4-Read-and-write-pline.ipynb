# Read and write .pline files
.pline files are text files format for Aspen SKUA (previously Gocad SKUA)

The function `networkx_to_pline` exports either with the edges or juste the nodes. It is possible to export properties that contains either 1 or 2 variables.


```python
import karstnet as kn
import networkx as nx
import numpy as np
```

```python
#load graph data

G = kn.io_func.networkx_from_yaml('../data/047_Seefeldhoele_000/047_Seefeldhoele_000.yaml')
```

```python

export_path = "../data/test_export_examples/"
cavename = 'Seefeldhoele'


# export the edges
kn.io_func.networkx_to_pline(G, data_type = 'lines-points',
                    properties = ['csdim'],                       
                    name = f'{export_path}{cavename}-lines'
                    )

# # export the nodes
# kn.io_func.networkx_to_pline(G, data_type = 'points-only', 
#                    properties = ['csdim'],
#                     name = f'{export_path}{cavename}-points'
#                     )
```
